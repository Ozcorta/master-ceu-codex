```python
import matplotlib.pyplot as plt
# SIMULACIÓN DOCENTE. No datos de A/B ni persona real.
sesion=[1,2,3,4,5,6,7,8]
fuerza_N=[210,215,212,218,221,225,224,230]
assert len(sesion)==len(fuerza_N)==8
assert sesion==sorted(set(sesion))
fig,ax=plt.subplots(figsize=(9,4.8))
ax.plot(sesion[:3],fuerza_N[:3],'o-',color='#585254',label='Fase A: observación')
ax.plot(sesion[3:],fuerza_N[3:],'o-',color='#397866',label='Fase B: programa ficticio')
ax.axvline(3.5,color='#a50018',linestyle='--',linewidth=1)
for x,y in zip(sesion,fuerza_N):
    ax.annotate(str(y),(x,y),xytext=(0,7),textcoords='offset points',ha='center')
ax.set(xlabel='Sesión',ylabel='Fuerza isométrica de press (N)',title='Seguimiento ficticio: descripción, no efecto causal',xticks=sesion,ylim=(0,260))
ax.grid(axis='y',alpha=.2)
ax.legend(loc='lower right')
fig.text(.01,.01,'SIMULACIÓN DOCENTE · 8 observaciones · No datos de los artículos A/B',fontsize=9)
fig.tight_layout(rect=(0,.04,1,1))
fig.savefig('figura_ficticia.png',dpi=180)
plt.show()
```
